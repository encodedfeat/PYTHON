```python
# creating stack using python list and taking assumption that we are using fixed size array(general array) like we dont have python list we have java or c array
```

```python
class Stack:
    def __init__(self,size):
        self.size=size
        self.stack=[None]*self.size
        self.top=-1
    def push(self,value):
        if self.top== self.size-1:
            return "overflow"
        else:
            self.top+=1
            self.stack[self.top]=value
    def pop(self):
        if self.top==-1:
            return "empty"
        else:
            data=self.stack[self.top]
            self.top-=1
            print(data)

    def traverse(self):
        for i in self.stack:
            print(i, end=" ")
```

```python
s=Stack(3)
```

```python
s.push(4)
```

```python
s.traverse()
```
