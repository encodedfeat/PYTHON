# implementing python list using c array(static, referential but not dynamic by default) and using ctypes module


```python
import ctypes
```

```python
class myList:
    def __init__(self):
        self.size=1
        self.n=0
        #creating a ctype array with size= self.size
        self.A=self.__create_array(self.size)

    #len of the list
    def __len__(self):
        return self.n
    #append 

    def append(self,item):
        if self.size==self.n:
            #resize
            self.__resize(self.size*2)
        #append
        self.A[self.n]=item
        self.n=self.n +1
        
    def __resize(self,new_capacity):
        B= self.__create_array(new_capacity)
        self.size=new_capacity
        for i in range(self.n):
            B[i]=self.A[i]
        #reinitialising because all other code works for A not B 
        self.A= B
    #printing
    def __str__(self):
        result =''
        for i in range(self.n):
            result=  result + str(self.A[i]) + "," 
        return "[" + result[:-1] +"]"

    #indexing 
    def __getitem__(self,index):
        if 0<=index<self.n:
             return self.A[index]

        return "index error-index out of range"

    #pop
    def pop(self):
        if self.n==0:
         return "Empty List"
        print(self.A[self.n-1])
        self.n=self.n-1

    #clear

    def clear(self):
        self.size=1
        self.n=0

    #find

    def find(self,item):
        for i in range(self.n):
            if self.A[i]==item:
                return i
        return "valueerror - item not in the list"

    #insert 

    def insert(self,index,item):
        if self.n==self.size:
            __resize(self.size*2)

        for i in range(self.n,index,-1):
            self.A[i]=self.A[i-1]
        self.A[index]=item
        self.n=self.n +1

    #delete 

    def __delitem__(self,index):
        if 0<=index<self.n:
            for i in range(index+1,self.n):
                self.A[i-1]=self.A[i]
            self.n=self.n-1 

    #remove
    def remove(self,item):
        index=self.find(item)
        if type(index)==int:
            self.__delitem__(index)
        else:
            return index
    def __create_array(self,capacity):
        #creates a ctypes array:
        return (capacity*ctypes.py_object)()
```
