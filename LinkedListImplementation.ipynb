```python
#now creating the linkedlist class with existing node class
class Node:
    def __init__(self,value):
        self.data=value
        self.next=None 
        
class LinkedList:
    def __init__(self):
        #empty linked list
        self.head=None 
        #number of nodes
        self.n=0

    #length of linked list
    def __len__(self):
        return self.n

    #traverse or print
    def __str__(self):
        curr=self.head
        result =""
        while curr!=None:
            result = result + str(curr.data) + "->"
            curr=curr.next
        return result[:-2]

    #insert head
    def insert_head(self,value):
        #creating a new node
        newNode=Node(value) 
        #makin connection
        newNode.next=self.head
        self.head=newNode
        self.n+=1

    
    def append(self,value):
        newNode=Node(value)
        if self.head==None:
            self.head=newNode
            self.n+=1
            return
        curr =self.head
        while curr.next!=None:
            curr=curr.next
        #reached at the last node 
        curr.next=newNode
        self.n+=1

    #insert at the position 0th,1st,2nd 3rd to n
#can be done using single pointer but was fun doing 2 pointer
    def insert_pos(self,pos,value):
        newNode=Node(value)
        if not 0 <= pos <= self.n:
            return "Invalid position"
        if (self.n==0 or pos==0):
            return self.insert_head(value)
        if pos==self.n:
            return self.append(value)
        
        if 0<=pos<=self.n:    
            prev=None
            next1=self.head
            count=0
            while count<pos:
                prev=next1
                next1=next1.next
                count+=1
            prev.next=newNode
            newNode.next=next1
            self.n+=1
        else: return "error"

    # insert after particular element

    def insert_after(self,element,value):
        newNode=Node(value)
        curr=self.head
        while curr !=None:
            if curr.data==element:
                break
            curr=curr.next
        if curr !=None:    
            newNode.next=curr.next
            curr.next=newNode
            self.n+=1
        else:
            return "element not found"


    #clearing the linkedlist:
    def clear(self):
        
        self.head=None
        self.n=0

    #delete from head:

    def delete_head(self):
        if self.head == None:
            return "Empty Linked List"
        self.head=self.head.next
        self.n=self.n-1

    #delete from tail

    def pop(self):
        curr=self.head
        if self.head is None:
            return "Empty List"
        if curr.next==None:
            return self.delete_head()
        
        while curr.next.next != None:
            curr=curr.next
            
        curr.next=None
        self.n=self.n-1

    #delete by item 

    def remove(self,item):
        if self.head==None:
            return "Empty Linked List"
        if self.head.data==item:
                return self.delete_head()
    
        curr=self.head
        while curr.next != None:
            if curr.next.data==item:
                break
            curr=curr.next
        if curr.next ==None:
            return "Not found"

        else:
            curr.next=curr.next.next
            self.n=self.n-1
            

            
            
            
```
