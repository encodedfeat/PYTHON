```python
class Node:
    def __init__(self, value):
        self.data=value
        self.next=None

class Stack:
    def __init__(self):
        self.top=None
        self.n=0

    def isempty(self):
        return self.top==None

    def push(self,value):
        new_cell=Node(value)
        new_cell.next=self.top
        self.top=new_cell
        self.n+=1

    def pop(self):
        if self.top==None:
            return 'empty stack'
        pop=self.top.data
        self.top=self.top.next
        self.n-=1
        return pop

    def peek(self):
        if self.top==None:
            return 'stack is empty'
        return self.top.data

    def size(self):
        return self.n
        
    def __str__(self):
        curr=self.top
        result =''
        while curr !=None:
            result =  result + str(curr.data) + ","
            curr=curr.next

        return ('[' + result[:-1] +']')

    
        
    
        
        
```

```python
def cel(M):
        s=Stack()
        n=len(M)
        for i in range(n):
            s.push(i)
       
        while s.size()>=2:
            i=s.pop()
            j=s.pop()

            if M[i][j]==0:
                s.push(i)
            else:
                s.push(j)

        cel=s.pop()

        for i in range(n):
            if cel != i:
                if M[i][cel]==0 or M[cel][i]==1:
                    print("no cel")
                    return 
        print("cel is ",cel)
        return 
```
